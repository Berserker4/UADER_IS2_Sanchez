# TP10 - Ejercicio 1 (VPN y rentabilidad)

Inversion: 1000/mes vencido. Beneficio al entregar: VP=18000 en ese momento. r=1% mensual.

- VPN = VP(beneficio) - VP(inversion)
- Rentabilidad = VPN / VP(inversion)


```python
from ej1_vpn import escenario, pv, R, BENEFICIO, COSTO_MENSUAL

plan = escenario(12)
ext15 = escenario(15)
ext18 = escenario(18)
pm = escenario(12, COSTO_MENSUAL * 1.05)

ann15 = (1 - (1 + R) ** (-15)) / R
c_max = (pv(BENEFICIO, 15) - plan["vpn"]) / ann15

rows = [
    ("a) Plan 12 meses", plan),
    ("b) Extension 15 meses", ext15),
    ("d) Retraso 6 meses (18)", ext18),
    ("e) PM +5% costo, a tiempo", pm),
]
for nombre, esc in rows:
    print(nombre)
    print(f"  VP inversion = {esc['vp_inversion']:.2f}")
    print(f"  VP beneficio = {esc['vp_beneficio']:.2f}")
    print(f"  VPN          = {esc['vpn']:.2f}")
    print(f"  Rentabilidad = {esc['rentabilidad']*100:.2f}%")
    print()

print(f"f) Costo mensual max (15m, mismo VPN plan): {c_max:.2f}")
print(f"e) Conviene PM vs retraso 3m? {pm['vpn']:.2f} > {ext15['vpn']:.2f} -> {pm['vpn'] > ext15['vpn']}")
```

## Interpretacion breve

- El retraso alarga la inversion y posterga el beneficio: el VPN cae fuerte.
- Con 6 meses de demora el VPN se vuelve negativo.
- Pagar +5% por gestion profesional conviene financieramente frente a demorar 3 meses.
